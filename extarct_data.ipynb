```python
import requests
import fitz  # PyMuPDF
from bs4 import BeautifulSoup

def fetch_content(url):
    response = requests.get(url)
    response.raise_for_status()  # Ensure we notice bad responses
    content_type = response.headers.get('Content-Type')
    return response.content, content_type

def extract_pdf_content(pdf_data):
    pdf_text = ""
    pdf_document = fitz.open(stream=pdf_data, filetype="pdf")
    for page_num in range(pdf_document.page_count):
        page = pdf_document.load_page(page_num)
        pdf_text += page.get_text()
    return pdf_text

def extract_webpage_content(html_data):
    soup = BeautifulSoup(html_data, 'html.parser')
    return soup.get_text(separator=' ', strip=True)

def extract_plain_text(content):
    return content.decode('utf-8')

def extract_data_from_url(url):
    content, content_type = fetch_content(url)
    if 'application/pdf' in content_type:
        return extract_pdf_content(content)
    elif 'text/html' in content_type:
        return extract_webpage_content(content)
    elif 'text/plain' in content_type:
        return extract_plain_text(content)
    else:
        return "Unsupported content type: {}".format(content_type)

# Example URL - Replace with your URL
url = input("Enter URL:")  # Change this to your desired URL
data = extract_data_from_url(url)
print(data)
```
